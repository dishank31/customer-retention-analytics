## 1. Run Churn Prediction Pipeline

```python
import sys
import os
import pandas as pd
import matplotlib.pyplot as plt

sys.path.append(os.path.abspath('..'))
import config
from src import churn_model

profitable = pd.read_csv(os.path.join(config.DATA_PROCESSED, 'customer_profitable.csv'))
transactions = pd.read_csv(os.path.join(config.DATA_RAW, 'transactions.csv'))

churn_data, models, best_model_name, model_results = churn_model.run(
    customer_data=profitable,
    transactions=transactions
)
```

## 2. Model Comparison

```python
results_df = pd.DataFrame(model_results).T
print(f"\n★ Best Model: {best_model_name}\n")

# Visual comparison
fig, ax = plt.subplots(figsize=(12, 6))
results_df[['Accuracy', 'Precision', 'Recall', 'F1', 'AUC-ROC']].plot(
    kind='bar', ax=ax, colormap='viridis', edgecolor='white', width=0.8
)
ax.set_title('Model Performance Comparison', fontsize=14, fontweight='bold')
ax.set_ylabel('Score')
ax.tick_params(axis='x', rotation=0)
ax.legend(loc='lower right')
ax.set_ylim(0, 1.05)
plt.tight_layout()
plt.show()

display(results_df.round(3))
```

## 3. Churn Distribution Analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Churn probability distribution
axes[0].hist(churn_data['churn_prob'], bins=50, color='#e74c3c', edgecolor='white', alpha=0.8)
axes[0].axvline(0.5, color='black', linestyle='--', linewidth=2, label='Threshold')
axes[0].set_title('Churn Probability Distribution', fontweight='bold')
axes[0].set_xlabel('Churn Probability')
axes[0].legend()

# Churn by segment
seg_churn = churn_data.groupby('rfm_segment')['churn_prob'].mean().sort_values(ascending=False)
colors = ['#e74c3c' if v > 0.5 else '#f39c12' if v > 0.3 else '#2ecc71' for v in seg_churn.values]
seg_churn.plot(kind='barh', ax=axes[1], color=colors)
axes[1].set_title('Avg Churn Probability by Segment', fontweight='bold')
axes[1].axvline(0.5, color='red', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

print(f"\nOverall predicted churn rate: {churn_data['churn_predicted'].mean():.1%}")
print(f"Customers predicted to churn: {churn_data['churn_predicted'].sum():,} / {len(churn_data):,}")
```
